Given an integer array `nums` and an integer `val`, remove **all occurrences** of `val` in `nums` **in-place**. The order of the elements may be changed. Then return the number of elements in `nums` which are **not equal** to `val`.

Consider the number of elements in `nums` which are not equal to `val` be `k`. To get accepted, you need to do the following things:

1. Change the array `nums` such that the first `k` elements of `nums` contain the elements which are not equal to `val`.
2. The remaining elements of `nums` are not important.
3. Return `k`.

Example 1
Input:
nums = [3, 2, 2, 3]
val = 3

Output:
k = 2
nums = [2, 2, _, _]
（順序不重要，底線表示多餘的資料）

說明：
移除所有的 3，剩下兩個 2。

Example 2
Input:
nums = [0, 1, 2, 2, 3, 0, 4, 2]
val = 2

Output:
k = 5
nums = [0, 1, 4, 0, 3, _, _, _]

說明：
移除所有的 2，剩下的五個元素為 [0, 1, 4, 0, 3]（順序可變）。

---

```python
def removeElement(nums,val):
  point=0
  for n in nums:
    if n!=val:
      nums[point]=n
      point+=1
  return point

nums=[3,2,2,3]
val=3
print(removeElement(nums,val))
```
